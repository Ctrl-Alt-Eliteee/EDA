```python
from sklearn.preprocessing import OneHotEncoder
encode = OneHotEncoder(sparse_output=False)
encoded = encode.fit_transform(df[['furnishingstatus']])
edf=pd.DataFrame(encoded,columns=encode.get_feature_names_out())
df=df.drop(['furnishingstatus'],axis=1)
df=pd.concat([df,edf],axis=1)
```

```python
ndf=pd.read_csv('processed-housepricing.csv')
```

```python
ndf
```

```python
ndf
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.ensemble import GradientBoostingRegressor,RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
```

```python
#drop the 'unnamed: 0' column which is causing the data leak 
x = ndf.drop(columns=['price','Unnamed: 0'])
y = ndf['price']
#Re-split the data
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
#Rescale the area feature
x_train[['area']]=scaler.fit_transform(x_train[['area']])
x_test[['area']]=scaler.transform(x_test[['area']])
print("Data leak fixed by removing 'Unnamed: 0'.Running models...")
models={
    'Linear Regression' : LinearRegression(),
    'Ridge':Ridge(random_state=42),
    'Lasso':Lasso(random_state=42),

    'Random Forest Regressor':RandomForestRegressor(random_state=42),
    'Gradient Boosting Regressor':GradientBoostingRegressor(random_state=42),

    'KNN':KNeighborsRegressor()
}
#Re-run evaluations
for name, model in models.items():
    try:
        model.fit(x_train,y_train)
        y_pred=model.predict(x_test)
        print(f'Model Name:{name}')
        print(f'R2 Score:{r2_score(y_test,y_pred):.3f}')
        print(f'MAE:{mean_absolute_error(y_test,y_pred):.3f}\n')
    except Exception as e:
        print(f'Model Name: {name}Failed: {e}')
    
```
